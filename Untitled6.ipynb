```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("cybersecurity_cases_india_combined.csv")
print(df.head(10))
```

```python
print(df['Category'].value_counts())
```

```python
q1 = df['Amount_Lost_INR'].quantile(0.25)
q3 = df['Amount_Lost_INR'].quantile(0.75)
iqr = q3 - q1
low = q1 - 1.5 * iqr
high = q3 + 1.5 * iqr
outliers = df[(df['Amount_Lost_INR'] < low) | (df['Amount_Lost_INR'] > high)]
print("Total Outliers:", len(outliers))
```

```python
print(df['Amount_Lost_INR'].sum())
```

```python
df['Incident_Type'].value_counts().plot(kind='bar')
plt.title("Incident Types")
plt.show()
```

```python
df['Category'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.show()
```

```python
df.groupby('Category')['Amount_Lost_INR'].sum()
```

```python
city_cases = df['City'].value_counts().head(10)
plt.figure(figsize=(8,5))
city_cases.plot(kind='bar')
plt.title("Top 10 Cities by Cyber Incidents")
plt.xlabel("City")
plt.ylabel("Number of Incidents")
plt.xticks(rotation=45)
plt.show()
```

```python
loss_by_city = df.groupby('City')['Amount_Lost_INR'].sum()
loss_by_city.plot(kind='bar')
plt.title("Financial Loss by City")
plt.xlabel("City")
plt.ylabel("Amount Lost")
plt.show()
```

```python
df['Year'].value_counts().sort_index().plot(kind='line')
plt.title("Cases by Year")
plt.show()
```
